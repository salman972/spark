```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

data = [
    (1, "salman", 78),
    (2, "gudly", 45),
    (3, "shreyank", 96)
]

df1 = spark.createDataFrame(
    data,
    ["id", "name", "marks"]
)
#display(df1)
df1.printSchema()

schema = StructType([
    StructField("id", IntegerType()),
    StructField("name", StringType()),
    StructField("marks", IntegerType())
])

df = spark.createDataFrame(
    data,
    schema
)
#display(df)
df.printSchema()
```

```python
df= df.filter(col("marks") == 96)
df.select(col("id")).show()
```

```python
df.explain()
```

```python
data1 = [(i, f"product_{i}", f"category_{i%5}") for i in range(1, 21)]
data2 = [(i, f"category_{i%5}", f"region_{i%3}") for i in range(1, 21)]

df_a = spark.createDataFrame(data1, ["product_id", "product_name", "category"])
df_b = spark.createDataFrame(data2, ["product_id", "category", "region"])
```

```python
df_a.show()
```

```python
df_a.join(df_b,df_a["product_id"] == df_b["product_id"],"inner").show()
```
